```python
import pandas as pd
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target  #房价中位数作为target
name_map = {
    'MedInc': '收入中位数',
    'HouseAge': '房龄中位数',
    'AveRooms': '平均房间数',
    'AveBedrms': '平均卧室数',
    'Population': '人口',
    'AveOccup': '平均入住人数',
    'Latitude': '纬度',
    'Longitude': '经度',
    'MedHouseVal': '房价中位数(10万美元)'
}

# 替换列名
df.rename(columns=name_map, inplace=True)

df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import os

# 图片输出目录
OUTPUT_DIR = "figures"
os.makedirs(OUTPUT_DIR, exist_ok=True)

plt.figure(figsize=(12, 8))
plt.rcParams['font.sans-serif'] = ['SimHei'] #黑体中文
plt.rcParams['axes.unicode_minus'] = False

features = ["收入中位数", "房龄中位数", "平均房间数", "平均卧室数", "人口", "房价中位数(10万美元)",]
n = len(features)
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(14, 8))
axes = axes.flatten()  # 展平成 1D 数组，方便遍历

for i, feat in enumerate(features):
    sns.kdeplot(data=df, x=feat, ax=axes[i], fill=True)
    axes[i].set_title(feat)

plt.suptitle("各特征密度分布", fontsize=14, y=1.02)
plt.tight_layout()

# 保存图片
plt.savefig(os.path.join(OUTPUT_DIR, "01_各特征密度分布.png"), dpi=150, bbox_inches="tight")
plt.show()
```

```python
df[['平均房间数', '平均卧室数','人口']].describe()
```

```python
# 只要有一个条件满足就显示出来
mask = (df["平均房间数"] > 80) | (df["人口"] > 9000) | (df["平均卧室数"] > 10)
df[mask]
```

```python

df_clean = df[
    (df["平均房间数"] <= 80) & 
    (df["平均卧室数"] <= 8) & 
    (df["人口"] <= 20000)
]

features = ["收入中位数", "房龄中位数", "平均房间数", "平均卧室数", "人口", "房价中位数(10万美元)",]
n = len(features)
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(14, 8))
axes = axes.flatten()  # 展平成 1D 数组，方便遍历

for i, feat in enumerate(features):
    sns.kdeplot(data=df_clean, x=feat, ax=axes[i], fill=True)
    axes[i].set_title(feat)

plt.suptitle("预处理后各特征密度分布", fontsize=14, y=1.02)
plt.tight_layout()

# 保存图片
plt.savefig(os.path.join(OUTPUT_DIR, "02_预处理后各特征密度分布.png"), dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df_clean.plot(
    kind="scatter",
    x="经度",
    y="纬度",
    alpha=0.2,
    s=df_clean["人口"] / 100,
    c="房价中位数(10万美元)",
    cmap="plasma",
    colorbar=True,
)
plt.title("加州房价地理分布")
plt.xlabel('经度')
plt.ylabel('纬度')

plt.tight_layout()

# 保存图片
plt.savefig(os.path.join(OUTPUT_DIR, "03_加州房价地理分布.png"), dpi=150, bbox_inches="tight")
plt.show()
```

```python
# 完整热力图
plt.figure(figsize=(10, 8))
sns.heatmap(df_clean.corr(), annot=True, cmap="coolwarm", fmt=".2f",
            vmin=-1, vmax=1, square=True)
plt.title("特征相关性热力图")
plt.tight_layout()

# 保存图片
plt.savefig(os.path.join(OUTPUT_DIR, "04_特征相关性热力图.png"), dpi=150, bbox_inches="tight")
plt.show()
```

```python
# 选几个关键特征 + 房价
cols = ["房价中位数(10万美元)", "收入中位数", "房龄中位数", "平均房间数", "纬度"]
g = sns.pairplot(df_clean[cols], diag_kind="kde", plot_kws={"alpha": 0.3, "s": 10})
g.figure.suptitle("关键特征散点图矩阵", y=1.02)

# 保存图片
g.savefig(os.path.join(OUTPUT_DIR, "05_关键特征散点图矩阵.png"), dpi=150, bbox_inches="tight")

plt.show()
```
